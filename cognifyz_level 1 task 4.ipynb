```python
#Task 3
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("Dataset cognifyz.csv")
df = df.dropna(subset=['Price range'])
df['Price range'] = df['Price range'].astype(int)
price_counts = df['Price range'].value_counts().sort_index()
print("Restaurants in each price range:\n")
print(price_counts)
total = len(df)
price_percentage = (price_counts / total) * 100
print("\nPercentage of restaurants in each price range:\n")
print(price_percentage.round(2))
plt.figure(figsize=(8,5))
price_counts.plot(kind='bar')
plt.title("Distribution of Restaurant Price Ranges")
plt.xlabel("Price Range Category")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=0)
plt.show()
plt.figure(figsize=(8,5))
plt.hist(df['Price range'], bins=4)
plt.title("Histogram of Price Ranges")
plt.xlabel("Price Range")
plt.ylabel("Frequency")
plt.show()
```

```python
#Task 4
import pandas as pd
delivery_counts = df['Has Online delivery'].value_counts()
delivery_percentage = (delivery_counts / len(df)) * 100
print("Online Delivery Percentage:\n")
print(delivery_percentage)
avg_rating_delivery = df.groupby('Has Online delivery')['Aggregate rating'].mean()
print("\nAverage Rating Comparison:\n")
print(avg_rating_delivery)
import matplotlib.pyplot as plt
avg_rating_delivery.plot(kind='bar', figsize=(6,4))
plt.title("Average Rating: Online Delivery vs No Delivery")
plt.xlabel("Online Delivery Available")
plt.ylabel("Average Rating")
plt.xticks(rotation=0)
plt.show()
```
